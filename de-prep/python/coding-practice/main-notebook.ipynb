Count word frequency in a text and return the top N words.

```python
from collections import Counter
import re

text = "The live-coding problems DE interviewers actually ask : what the fuck did you do all these years? What did you learn?"
def top_N_words(text, n):
    text = re.findall(r"[a-z]+", text.lower())  # We are taking the text and splitting the words and removing all the special character and including only alphabets
    print(Counter(text).most_common(n))

def top_N_words_no_counter(text, n):
    counts = {}
    for i in re.findall(r"[a-z]+", text.lower()):
        counts[i] = counts.get(i,0)+1  #get(key, default_value) avoids if the key is not found...
    print(sorted(counts.items(), key=lambda x : -x[1])[:n])
# top_N_words(text, 3)
top_N_words_no_counter(text, 3)
```

Reverse a string and check if it's a palindrome (ignoring case and
punctuation).

```python
string = "42ma-%&^&#^dam24"
def palindrome_check(text):
    corrected = [string.lower() for string in text if string.isalnum()]
    return corrected == corrected[::-1]

assert palindrome_check(string)
assert palindrome_check("A man, a plan, a canal: Panama") #testing 
```

Find duplicates in a list, then remove duplicates while preserving order.

```python
from random import randint
from collections import Counter

list1 = [randint(0,5) for _ in range(0,11)]
print(list1)
def find_dups(listt):
    return [x for x,count in Counter(listt).items() if count > 1]

def de_dup(listt):
    return list(dict.fromkeys(listt))

def manual_de_dup(listt):
    seen, out = set(), []
    for i in listt:
        if i not in seen: 
            seen.add(i)
            out.append(i)
    return out


print(f"fromkeys : {list(dict.fromkeys(list1,0))}")  # used to create a dictionary with list of keys
print(f"duplicates : {find_dups(list1)}")
print(f"no dups : {de_dup(list1)}")
print(f"manual method : {manual_de_dup(list1)}")
```

Two Sum: return the indices of two numbers that add up to a target

```python
from collections import Counter

def two_sum(nums: list[int], key=5):
    for ind,val in enumerate(nums):
        if (key-val) in nums: 
            if nums.index(key-val) != ind:
                return ind, nums.index(key-val)

def two_sum_with_dict(nums, key=5):
    seen = {}
    for i,v in enumerate(nums):
        comp = key-v
        if comp in seen:
            return i, seen[comp]
        seen[v] = i

list1 = [randint(0,5) for _ in range(0,11)]
print(list1)
print(two_sum(list1, 6))
print(two_sum_with_dict(list1, 6))
```
