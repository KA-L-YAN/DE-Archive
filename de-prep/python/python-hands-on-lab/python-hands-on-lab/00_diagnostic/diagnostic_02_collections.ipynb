## Question 1 — 🟢 Easy · Inverted index

**Implement:** `inverted_index(docs: dict[str, str]) -> dict[str, list[str]]`

Map each lowercase word to the sorted list of doc ids containing it (each id once). Keys sorted.

**Input:** `docs: dict[str, str]`

**Expected output:** `dict[str, list[str]]`

**Constraints:** Standard library only, unless the statement says otherwise.

**Example**

```python
inverted_index({'d2': 'Spark is fast', 'd1': 'python is FUN fun'})
# -> {'fast': ['d2'], 'fun': ['d1'], 'is': ['d1', 'd2'], 'python': ['d1'], 'spark': ['d2']}
```

**Implement:** `top_k(items: list[str], k: int) -> list[tuple[str, int]]`

**Input:** `items: list[str], k: int`

**Expected output:** `list[tuple[str, int]]`

**Example**

```python
top_k(['b', 'a', 'c', 'b', 'a', 'd'], 3)
# -> [('a', 2), ('b', 2), ('c', 1)]
```

**Implement:** `merge_intervals(intervals: list[tuple[int, int]]) -> list[tuple[int, int]]`

**Input:** `intervals: list[tuple[int, int]]`

**Expected output:** `list[tuple[int, int]]`
