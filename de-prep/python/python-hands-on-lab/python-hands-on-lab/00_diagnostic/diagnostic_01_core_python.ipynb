## Question 1 — 🟢 Easy · Parse a duration

**Implement:** `parse_duration(text: str) -> int`

Convert strings like `"1h30m15s"`, `"45m"`, `"2h"`, `"90s"` (units in this order, each optional, at least one present, case-insensitive) into total seconds. Anything else (`""`, `"1x"`, `"30s1h"`, `"h"`) → `ValueError`.

**Input:** `text: str`

**Expected output:** `int`

**Constraints:** Standard library only, unless the statement says otherwise.

**Example**

```python
parse_duration('1h30m15s')
# -> 5415
```

**Implement:** `compare_versions(a: str, b: str) -> int`

**Input:** `a: str, b: str`

**Expected output:** `int`

**Constraints:** Standard library only, unless the statement says otherwise.

**Example**

```python
compare_versions('1.10.0', '1.9')
# -> 1
```

## Question 4 — 🟠 Hard · Word-wrap for log messages

**Implement:** `wrap(text: str, width: int) -> list[str]`

Wrap whitespace-separated words into lines of at most `width` characters (single spaces between words). Words longer than `width` are split into `width`-sized pieces. `width < 1` → `ValueError`. Empty/blank text → `[]`.

**Input:** `text: str, width: int`

**Expected output:** `list[str]`

**Constraints:** Standard library only, unless the statement says otherwise.

**Example**

```python
wrap('the quick brown fox jumps', 10)
# -> ['the quick', 'brown fox', 'jumps']
```
