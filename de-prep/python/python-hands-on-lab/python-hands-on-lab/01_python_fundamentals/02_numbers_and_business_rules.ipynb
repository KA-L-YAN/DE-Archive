## Exercise 3 — 🟡 Medium · Shipping fee rule table

**Implement:** `shipping_fee(weight_kg: float, zone: str, express: bool = False) -> int`

Compute the fee in rupees from this table (zone is case-insensitive):

| zone | base (≤ 1 kg) | per extra kg (started) |
|---|---|---|
| local | 40 | 10 |
| national | 80 | 25 |
| international | 900 | 300 |

Extra weight is charged per *started* kg (1.2 kg → 1 extra kg; 2.0 kg → 1 extra kg). Express doubles the fee.
Orders `> 30 kg` are rejected with `ValueError`, as are `weight_kg <= 0` and unknown zones.

**Input:** `weight_kg: float, zone: str, express: bool = False`

**Expected output:** `int`

**Constraints:** Represent the table as data (a dict), not as an `if/elif` chain.

**Example**

```python
shipping_fee(1, "local")              # 40
shipping_fee(1.2, "National")         # 105
shipping_fee(3, "local", express=True) # 120
```

## Exercise 4 — 🟠 Hard · Transaction risk classification

**Implement:** `classify_risk(txn: dict) -> str`

Classify a card transaction as `"HIGH"`, `"MEDIUM"` or `"LOW"`. Fields: `amount` (float, INR-equivalent),
`currency` (str), `hour` (0–23, local time), `country` (str), `home_country` (str), `failed_attempts` (int, may be missing → 0).

First matching rule wins:
1. HIGH if `amount > 200000`, or `failed_attempts >= 3`.
2. HIGH if the country differs from `home_country` **and** `amount > 50000`.
3. MEDIUM if `0 <= hour < 5` and `amount > 10000`.
4. MEDIUM if the currency isn't `"INR"`, or the country differs from home.
5. Otherwise LOW.

Country/currency comparisons are case-insensitive. Missing `amount` or `hour` → `KeyError` is fine.

**Input:** `txn: dict`

**Expected output:** `str`

**Constraints:** Standard library only, unless the statement says otherwise.

**Example**

```python
classify_risk({"amount": 500, "currency": "INR", "hour": 14, "country": "IN", "home_country": "in"})  # 'LOW'
```

## Challenge — 🔴 Very Hard · Parse messy money strings

**Implement:** `parse_amount(text: str) -> Decimal`

Finance exports contain amounts like `"₹1,23,456.78"`, `"$1,234.50"`, `"1.234,50 €"` (EU style),
`"(1,200.00)"` (accounting negative), `"-45"`, `" 12 "`, `"USD 99.9"`. Return a `Decimal` (exact, not rounded).

Rules: strip currency symbols/codes (₹ $ € £, INR USD EUR GBP) and spaces. Parentheses mean negative.
If both `,` and `.` appear, the one that appears **last** is the decimal separator. If only `,` appears and it's
followed by exactly 2 digits at the end, it's a decimal separator; otherwise it's a thousands separator.
Anything left that isn't a valid number → `ValueError`. Empty → `ValueError`.

**Input:** `text: str`

**Expected output:** `Decimal`

**Constraints:** Don't use `float` at any point.

**Example**

```python
parse_amount("1.234,50 €")    # Decimal('1234.50')
parse_amount("(1,200.00)")    # Decimal('-1200.00')
parse_amount("12,5")          # Decimal('125')  - one digit after the comma => thousands separator
```
