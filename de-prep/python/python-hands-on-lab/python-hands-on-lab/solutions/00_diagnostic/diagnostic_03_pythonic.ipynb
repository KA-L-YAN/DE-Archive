## Question 1 — 🟢 Easy · Transpose

**Working solution**

```python
def transpose(matrix):
    return [list(col) for col in zip(*matrix)]
```

**Explanation**

`zip(*rows)` *is* transpose.

**Complexity:** O(cells).

**Common mistakes**

- Nested index loops.

## Question 2 — 🟡 Medium · First repeated element, lazily

**Working solution**

```python
def first_repeat(iterable):
    seen = set()
    for item in iterable:
        if item in seen:
            return item
        seen.add(item)
    return None
```

**Explanation**

A set for O(1) membership; returning inside the loop stops consuming the generator.

**Complexity:** O(n).

**Common mistakes**

- `list(iterable)` first — hangs on infinite input.

## Question 3 — 🟡 Medium · Group anagrams

**Working solution**

```python
def group_anagrams(words):
    groups = {}
    for w in words:
        groups.setdefault("".join(sorted(w.lower())), []).append(w)
    return list(groups.values())
```

**Explanation**

A canonical key (sorted letters) + an insertion-ordered dict.

**Complexity:** O(n · k log k).

**Common mistakes**

- Comparing every pair of words.

## Question 4 — 🟠 Hard · Running balance and first overdraft

**Working solution**

```python
from itertools import accumulate

def balances(opening, txns):
    ids = [t for t, _ in txns]
    running = [round(b, 2) for b in accumulate((amt for _, amt in txns), initial=opening)][1:]
    pairs = list(zip(ids, running))
    overdraft = next((t for t, b in pairs if b < 0), None)
    return pairs, overdraft
```

**Explanation**

`accumulate(..., initial=x)` yields the opening balance first; slicing it off aligns balances with txns. `next(gen, None)` = 'first match or None'.

**Complexity:** O(n).

**Common mistakes**

- Manual index bookkeeping.
