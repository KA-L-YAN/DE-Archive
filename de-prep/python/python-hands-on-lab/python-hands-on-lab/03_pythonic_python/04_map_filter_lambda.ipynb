## Challenge — 🔴 Very Hard · Generic group-and-reduce

**Implement:** `group_reduce(records, key, value, reducer, initial) -> dict`

A tiny `reduceByKey`: `key` and `value` are functions (or strings meaning "this field"). For each record, fold `value(record)`
into the group's accumulator with `reducer(acc, v)`, starting from `initial` — which may be a mutable value like `[]` or a
callable producing a fresh initial value (`initial=list`) so groups don't share state. Return groups in first-seen order.

**Input:** `records, key, value, reducer, initial`

**Expected output:** `dict`

**Constraints:** Standard library only, unless the statement says otherwise.

**Example**

```python
group_reduce(orders, "country", lambda o: o["quantity"], lambda a, v: a + v, 0)        # sum per country
group_reduce(orders, "country", "order_id", lambda a, v: a + [v], list)                   # collect ids
```
