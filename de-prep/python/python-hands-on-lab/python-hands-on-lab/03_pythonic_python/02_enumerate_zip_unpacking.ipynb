# enumerate, zip, Unpacking & Slicing

## What you'll practice

Index loops (`for (int i = 0; i < n; i++)`) almost never appear in idiomatic Python. `enumerate` gives you the index,
`zip` walks sequences in lockstep (`strict=True` catches length mismatches), `zip(*rows)` transposes, and unpacking
(`a, *rest = ...`) destructures. `itertools.accumulate`/`pairwise` cover running values.

## Exercise 4 — 🟠 Hard · Row-oriented ↔ column-oriented

**Implement:** `to_columnar(records: list[dict]) -> dict[str, list]`, `from_columnar(columns: dict[str, list]) -> list[dict]`

Parquet/Arrow/Spark store data by column. `to_columnar` converts records to `{column: [values...]}` — the union of keys
across records, in first-seen order, with `None` for missing values. `from_columnar` reverses it; columns of unequal length → `ValueError`.

**Input:** `records: list[dict]`

**Expected output:** `dict[str, list]`

**Constraints:** Standard library only, unless the statement says otherwise.

**Example**

```python
to_columnar([{"a": 1, "b": 2}, {"a": 3, "c": 4}])   # {'a': [1, 3], 'b': [2, None], 'c': [None, 4]}
```

## Challenge — 🔴 Very Hard · Fixed-width mainframe file parser

**Implement:** `parse_fixed_width(lines: list[str], spec: list[tuple[str, int, callable]]) -> list[dict]`

Banks still ship fixed-width files. `spec` is `[(name, width, converter), ...]` in field order. Compute start offsets
with `itertools.accumulate`, slice each field, `strip()` it, and convert (empty → `None`). Lines shorter than the total width are
padded with spaces; longer lines → `ValueError` with the 1-based line number. Fields named `"_"` are fillers: skip them.

**Input:** `lines: list[str], spec: list[tuple[str, int, callable]]`

**Expected output:** `list[dict]`

**Constraints:** Standard library only, unless the statement says otherwise.

**Example**

```python
spec = [("acct", 6, str), ("_", 2, str), ("amount", 10, float), ("ccy", 3, str)]
parse_fixed_width(["A00001  0000150.75INR"], spec)  # [{'acct': 'A00001', 'amount': 150.75, 'ccy': 'INR'}]
```
