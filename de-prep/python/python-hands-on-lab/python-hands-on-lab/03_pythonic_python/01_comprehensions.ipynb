# Comprehensions: list, dict, set, conditional, nested

## What you'll practice

Comprehensions replace most `for` + `append` loops and most Java stream pipelines. Rule of thumb: if it fits on ~2 lines
and reads left-to-right, use a comprehension; if you need `try/except` or several statements, use a helper function
(or a loop). You'll also meet the walrus operator `:=` for "compute once, filter and keep".

## Exercise 4 — 🟠 Hard · Schema-driven record normalization

**Implement:** `normalize(records: list[dict], schema: dict[str, callable], rename: dict[str, str] | None = None) -> list[dict]`

Apply a schema to raw records: output **only** schema columns, in schema order. `rename` maps raw column names to schema names
(`{"Order ID": "order_id"}`) and is applied first. Each value is converted with the schema's callable; a missing column, an empty
string, or a conversion error yields `None` for that field (don't drop the record).

**Input:** `records: list[dict], schema: dict[str, callable], rename: dict[str, str] | None = None`

**Expected output:** `list[dict]`

**Constraints:** Standard library only, unless the statement says otherwise.

**Example**

```python
normalize([{"Order ID": "7", "amt": "1.5", "junk": 1}], {"order_id": int, "amount": float}, {"Order ID": "order_id", "amt": "amount"})
# [{'order_id': 7, 'amount': 1.5}]
```

**Implement:** `pivot(rows: list[dict], index: str, columns: str, values: str, agg=sum, fill=0) -> dict[str, dict[str, float]]`

**Input:** `rows: list[dict], index: str, columns: str, values: str, agg=sum, fill=0`

```python
pivot([{"c": "IN", "ch": "web", "q": 2}, {"c": "IN", "ch": "web", "q": 3}, {"c": "FI", "ch": "app", "q": 1}], "c", "ch", "q")
# {'FI': {'app': 1, 'web': 0}, 'IN': {'app': 0, 'web': 5}}
```
