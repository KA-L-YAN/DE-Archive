## Challenge — 🔴 Very Hard · Flatten and unflatten JSON documents

**Implement:** `flatten_dict(doc: dict, sep: str = '.') -> dict`, `unflatten_dict(flat: dict, sep: str = '.') -> dict`

`flatten_dict` turns nested dicts/lists into one level: `{"a": {"b": 1}, "items": [{"x": 1}, {"x": 2}]}` →
`{"a.b": 1, "items.0.x": 1, "items.1.x": 2}`. Empty dicts/lists are kept as leaf values (`{"tags": []}` → `{"tags": []}`).

`unflatten_dict` reverses it. At each level, if **all** keys are the consecutive integers `0..n-1` (as strings), build a
list; otherwise a dict. `unflatten_dict(flatten_dict(d)) == d` for JSON-like data without digit-only dict keys.

**Input:** `doc: dict, sep: str = '.'`

**Expected output:** `dict`

**Constraints:** Standard library only, unless the statement says otherwise.

**Example**

```python
unflatten_dict({'a.0': 'x', 'a.2': 'y'})
# -> {'a': {'0': 'x', '2': 'y'}}
```
