## Exercise 4 — 🟠 Hard · Effective permissions with wildcards

**Implement:** `effective_permissions(roles: list[str], role_perms: dict[str, set[str]], denied: set[str]) -> set[str]`

A user has several roles. Their permissions are the **union** of their roles' permissions, minus `denied`.
Wildcards: a granted `"orders:*"` expands to every concrete permission starting with `"orders:"` that appears in **any** role
in `role_perms`; a denied `"orders:*"` removes all `orders:` permissions. Unknown roles are ignored. The result contains no wildcards.

**Input:** `roles: list[str], role_perms: dict[str, set[str]], denied: set[str]`

**Expected output:** `set[str]`

**Constraints:** Standard library only, unless the statement says otherwise.

**Example**

```python
role_perms = {"viewer": {"orders:read", "customers:read"}, "ops": {"orders:*"}, "admin": {"orders:delete", "orders:write"}}
effective_permissions(["viewer", "ops"], role_perms, {"orders:delete"})
# {'customers:read', 'orders:read', 'orders:write'}
```

## Challenge — 🔴 Very Hard · Group duplicate customers (connected components)

**Implement:** `duplicate_groups(customers: list[dict]) -> list[list[str]]`

(digits only, last 10 digits). Duplication is transitive: A~B by email and B~C by phone ⇒ {A, B, C}. Return groups of
`id`s with size > 1, each group sorted, groups sorted by first id. Missing/blank email or phone never matches.

**Input:** `customers: list[dict]`

**Expected output:** `list[list[str]]`

**Constraints:** Must handle 20k customers quickly — no O(n²) pairwise comparison.

**Example**

```python
duplicate_groups([
  {"id": "1", "email": "K@x.io", "phone": None},
  {"id": "2", "email": "k@x.io ", "phone": "+91 98765-43210"},
  {"id": "3", "email": None, "phone": "9876543210"},
  {"id": "4", "email": "other@x.io", "phone": ""}])
# [['1', '2', '3']]
```
