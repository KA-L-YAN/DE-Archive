**Implement:** `class Txn(NamedTuple)`, `parse_txns(rows: list[list[str]]) -> list[Txn]`

`is_debit(self) -> bool` (amount < 0). `parse_txns` converts raw CSV rows `[txn_id, account, amount, currency]` into `Txn`s,
uppercasing currency and skipping rows whose amount isn't numeric.

**Input:** see the class description above

**Expected output:** an instance behaving as described

**Constraints:** Standard library only, unless the statement says otherwise.

**Example**

```python
t = parse_txns([["T1", "A1", "-250.5", "inr"]])[0]
t.amount, t.currency, t.is_debit()   # (-250.5, 'INR', True)
t._replace(amount=0)                 # new Txn; t is unchanged
```

## Challenge — 🔴 Very Hard · Change data capture between two snapshots

**Implement:** `diff_snapshots(old: list[tuple], new: list[tuple], key_len: int = 1) -> tuple[list, list, list]`

Rows are tuples whose first `key_len` elements form the primary key (e.g. `("C001", "Kalyan", "Hyderabad")`).
Compare yesterday's and today's full extracts and return `(inserted, deleted, updated)`:
- `inserted`: new rows (key only in `new`), sorted by key
- `deleted`: old rows (key only in `old`), sorted by key
- `updated`: `(old_row, new_row)` pairs whose key matches but other columns differ, sorted by key

Duplicate keys within one snapshot → `ValueError`. This is exactly what an SCD/CDC job does.

**Input:** `old: list[tuple], new: list[tuple], key_len: int = 1`

**Expected output:** `tuple[list, list, list]`

**Constraints:** Standard library only, unless the statement says otherwise.

**Example**

```python
old = [("C1", "Kalyan", "Hyd"), ("C2", "Diya", "Pune")]
new = [("C1", "Kalyan", "Vijayawada"), ("C3", "Erik", "Espoo")]
diff_snapshots(old, new)
# ([('C3', 'Erik', 'Espoo')], [('C2', 'Diya', 'Pune')], [(('C1','Kalyan','Hyd'), ('C1','Kalyan','Vijayawada'))])
```
