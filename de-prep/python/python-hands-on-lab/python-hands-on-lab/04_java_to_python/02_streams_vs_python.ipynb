## Exercise 1 — 🟢 Easy · filter → map → collect

**Implement:** `big_spender_names(customers: list[dict], threshold: float) -> list[str]`

```java
customers.stream().filter(c -> c.getSpend() > threshold).map(Customer::getName).sorted().collect(Collectors.toList());
```

**Input:** `customers: list[dict], threshold: float`

**Expected output:** `list[str]`

**Constraints:** Standard library only, unless the statement says otherwise.

**Example**

```python
big_spender_names([], 0)
# -> []
```

```java
orders.stream().flatMap(o -> o.getTags().stream().distinct()).collect(groupingBy(identity(), counting()))
      .entrySet().stream().sorted(byValueDescThenKey).limit(n)...
```
