# Beans, Getters/Setters, Optional, Builders, Static Utils → Python

## What you'll practice

Python classes are lighter than Java's. Public attributes are normal; add a `@property` only when you need logic. `@dataclass` replaces
constructors/equals/hashCode/toString. `None` + `or`/`if x is None` replaces `Optional`. Keyword arguments with defaults replace builders.
Module-level functions replace `static` utility classes.
