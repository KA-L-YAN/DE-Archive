## Exercise 1 — 🟢 Easy · Build S3 paths from parts

**Implement:** `s3_path(bucket: str, *parts: str, filename: str | None = None) -> str`

Join into `s3://bucket/part1/part2/filename`, stripping stray `/` from every piece and skipping empty parts. `filename` is keyword-only by position (it comes after `*parts`).

**Input:** `bucket: str, *parts: str, filename: str | None = None`

**Expected output:** `str`

**Constraints:** Standard library only, unless the statement says otherwise.

**Example**

```python
s3_path("raw", "/sales/", "2026", filename="part-0.csv")   # 's3://raw/sales/2026/part-0.csv'
s3_path("raw")                                            # 's3://raw'
```

## Challenge — 🔴 Very Hard · A tiny command dispatcher driven by annotations · 🧩 Mini-project

**Implement:** `dispatch(command_line: str, handlers: dict[str, callable])`

Rules: first token = command (unknown → `ValueError` listing commands); `--name value` options (dashes → underscores); an option whose handler

values work (`--title "Daily Sales"`). Return the handler's result.

**Input:** `command_line: str, handlers: dict[str, callable]`

**Constraints:** Standard library only, unless the statement says otherwise.
