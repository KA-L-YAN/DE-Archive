## Exercise 4 — 🟠 Hard · Transactions with commit and rollback

**Implement:** `class KeyValueStore`

`KeyValueStore()` has `get(key, default=None)`, `set(key, value)`, `delete(key)` and `transaction()` — a context manager. Inside a transaction,
writes are **staged**: `get` sees staged values (read-your-writes), others reading `store.committed()` (a copy of committed data) don't. On normal exit the
staged changes commit atomically; on exception they're discarded and the exception propagates. Nested transactions → `RuntimeError`.
Count `store.commits` and `store.rollbacks`.

**Input:** see the class description above

**Expected output:** an instance behaving as described

**Constraints:** Standard library only, unless the statement says otherwise.

**Example:** the tests are executable examples — read `test_kv_commit, test_kv_rollback_and_nested` in `tests/functions_oop/test_05_05_context_managers.py`.

## Challenge — 🔴 Very Hard · Many files at once with ExitStack: k-way merge of sorted files · 🧩 Mini-project

**Implement:** `merge_sorted_files(paths: list, output_path, key_column: str) -> int`

Several CSV extracts are each sorted by `key_column` (e.g. `order_ts`). Merge them into one sorted CSV **without loading them into memory**:
open all inputs with `contextlib.ExitStack` (any open failure must close the already-opened files), stream rows through `heapq.merge(..., key=...)`,
write with `csv.DictWriter` using the first file's header. All inputs must share the same header (else `ValueError`, raised before writing anything).
Return the number of rows written. Empty input list → write nothing, return 0.

**Input:** `paths: list, output_path, key_column: str`

**Expected output:** `int`

**Constraints:** Standard library only, unless the statement says otherwise.
