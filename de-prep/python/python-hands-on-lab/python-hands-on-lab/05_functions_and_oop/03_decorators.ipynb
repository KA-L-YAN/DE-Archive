# Decorators

## What you'll practice

A decorator is a function that takes a function and returns a (usually wrapped) function. `@timed` is just `f = timed(f)`. Decorators with
arguments (`@retry(times=3)`) are *decorator factories*: a function returning a decorator. Always use `functools.wraps`.

## Exercise 4 — 🟠 Hard · Retry with exponential backoff

**Implement:** `retry(times=3, exceptions=(Exception,), delay=0.1, backoff=2.0, max_delay=30.0, sleep=time.sleep) -> callable`

`@retry(...)` re-invokes the function when it raises one of `exceptions`, up to `times` **attempts in total**. Sleep `delay`, then `delay*backoff`, ... capped at
`max_delay` between attempts (no sleep after the last attempt). Other exceptions propagate immediately. After the final failure re-raise the last exception.
Record the attempt count of the most recent call in `wrapper.last_attempts`. `times < 1` → `ValueError` at decoration time. `sleep` is injectable for tests.

**Input:** `times=3, exceptions=(Exception,), delay=0.1, backoff=2.0, max_delay=30.0, sleep=time.sleep`

**Expected output:** `callable`

**Constraints:** Standard library only, unless the statement says otherwise.

**Example:** the tests are executable examples — read `test_retry_succeeds_after_failures, test_retry_gives_up_and_reraises, test_retry_other_exception_immediate` in `tests/functions_oop/test_05_03_decorators.py`.
