# First-Class Functions & Closures

## What you'll practice

A closure is a function that remembers variables from the scope where it was created. It's Python's lightweight alternative to small
classes with one method: counters, configured validators, caches, rate limiters, event buses. `nonlocal` lets a closure rebind captured variables.

## Challenge — 🔴 Very Hard · An event bus from closures

**Implement:** `make_event_bus() -> tuple[callable, callable]`

Return `(subscribe, publish)`.
- `subscribe(event_type, handler)` registers a handler and returns an `unsubscribe()` function (calling it twice is harmless). `event_type="*"` receives every event.
- `publish(event_type, payload)` calls specific handlers first (in subscription order), then wildcard handlers, each as `handler(event_type, payload)`.
  A failing handler must not stop the others. Return `{"delivered": n_ok, "errors": [(handler_name, repr(exc)), ...]}`.
- Subscribing/unsubscribing **during** publish must not break the current delivery (iterate over a snapshot).

**Expected output:** `tuple[callable, callable]`

**Constraints:** Standard library only, unless the statement says otherwise.
