### List comprehension

**Takeaway:** `[expr for x in iterable]` builds a list in one line.

- `range(0, 11)` stops before 11, so `i*2` gives `0..20` (11 items).

```python
listt = [i*2 for i in range(0,11)]
print(listt)
```

### Shallow vs deep copy

**Takeaway:** `copy.copy` copies only the outer list; `copy.deepcopy` copies the nested lists too.

- Shallow copy shares the inner lists: `listt[2].append(333)` shows up in `shallow_copy` too.
- Deep copy is fully independent: `deep_copy[1].append(777)` does not touch `listt`.
- `shallow_copy[0] = 888` rebinds a slot in the copy only; `listt` is unchanged.


```python
import copy

listt = [[i*2] for i in range(0,11)]

deep_copy = copy.deepcopy(listt)
shallow_copy = copy.copy(listt)
listt[2].append(333)
shallow_copy[0] = 888
deep_copy[1].append(777)
print(listt)
print(deep_copy)
print(shallow_copy)
```

### Word frequency

**Takeaway:** lowercase, strip punctuation, split, then `Counter(...).most_common(n)`.

- `str.maketrans("", "", string.punctuation)` + `translate` deletes punctuation.
- `most_common(3)` returns `(word, count)` tuples, highest first.
- This cell imports `Counter`, which later cells reuse.

⚠️ Name typo (code unchanged): `word_ferquency`.


```python
from collections import Counter
import string

string_list = "Hey there !!! Pavan is calling another Pavan and this another pavan is looking at another pavan"

def word_ferquency(strr):
    strr2 = strr.lower().translate(str.maketrans("","",string.punctuation))
    return Counter(strr2.split()).most_common(3)
word_ferquency(string_list)
```

### Palindrome / anagram

- **Anagram:** same characters, same counts → `Counter(a) == Counter(b)`.
- ⚠️ `check_palindrom` compares `str1` with reverse of `str2`; a real palindrome check uses **one** string.
- ⚠️ `check_anagrams` counts spaces/punctuation (`'dormitory'` vs `'dirty room'` → `False`).
- Needs `Counter` from the word-frequency cell.


```python
def check_palindrom(str1, str2):
    return (str1.lower() == str2[::-1].lower())

def check_anagrams(str1, str2):
    return Counter(str1.lower()) == Counter(str2.lower())

print(check_anagrams('mama', 'papa'))
print(Counter('mama'))
```

### Generators

**Interview answer:** a generator produces values lazily, one at a time, so it is memory-efficient for large data.

- Function with `yield` = generator function; calling it returns a generator iterator.
- `yield` returns a value and pauses until the next one is asked for.
- ⚠️ Single-use. The `for` loop below calls `get_numbers()` again → a **new** generator, so it restarts at 1.


```python
def get_numbers():
    yield 1
    yield 2
    yield 3
    yield 4
numbers  = get_numbers()
print(next(numbers))
print(next(numbers))

for i in get_numbers():
    print(i)
```

### Generator expression

- `( ... )` instead of `[ ... ]`; lazy, one value at a time. Official term: _generator expression_ (not "generator comprehension").
- Printing shows only the object; consume with `next()`, `for`, `list()`, `sum()`.
- `next` is a built-in **function**, not a keyword.
- Here the `for` reuses the **same** generator, so it continues: `5 7 9`.


```python
even_numbers = [i for i in range(0,11) if i%2==0]
print(f"Normal list using comprehension : {even_numbers}")

odd_numbers = (i for i in range(0,11) if i%2!=0)
print(f"Generator : {odd_numbers}\n"+
      "Yes this wont print anything other than the object reference... you have to use next keyword to print")

print(next(odd_numbers))
print(next(odd_numbers))
for i in odd_numbers:
    print(i, end = ' ')
```

### Name mangling (`__name`)

**Purpose:** avoid name clashes in **subclasses**, not security.

- `self.__make` in `Car` becomes `_Car__make` (compile time); still reachable as `obj._Car__make`.
- Only names starting with `__` and not ending with `__`.
- `@property` = read-only getter; `@x.setter` adds validation.

⚠️ Setter line is a **SyntaxError** (assignment inside a conditional expression). Use a normal `if/else` and `raise ValueError`.
⚠️ `get_details()` returns a **set** (no `key: value` pairs), so order is random.


```python
class Car:
    def __init__(self, make, model, manufactured_year, on_road_price):
        self.__make = make
        self.__model = model
        self.__manufactured_year = manufactured_year
        self.__on_road_price = on_road_price

    @property
    def make(self):
        return self.__make
    @property
    def model(self):
        return self.__model
    @property
    def manufactured_year(self):
        return self.__manufactured_year
    @property
    def on_road_price(self):
        return self.__on_road_price
    
    @on_road_price.setter
    def on_road_price(self, price):
        print("Zero Value Error") if price <= 0 else self.__on_road_price = price

    def get_details(self):
        return {f"Make : {self.__make}",
                f"Model : {self.__model}",
                f"Year : {self.__manufactured_year}",
                f"Price : {self.__on_road_price}"}

new_car = Car('Tesla', 'roadster', '2027', 200_000.00)
print(new_car.get_details())
```

### Inheritance

**Takeaway:** `class Porsche(Vehicle)` → the child gets the parent's methods and properties (Java's `extends`).

- `super().__init__(...)` runs the parent's constructor so the parent's attributes get set. Unlike Java, it need not be the first line, but call it before using parent state.
- Parent's `self.__name` is mangled to `_Vehicle__name`, so the child can't write `self.__name` (it would become `_Porsche__name`). Go through the property (`self.name`).
- `__turbo` → `_Porsche__turbo`: no getter, so it is only reachable as `obj._Porsche__turbo`. It is also never used.
- `@x.setter` must come **after** the `@property` getter of the same name (order matters).
- `revv()` uses `self.name`, so it works for any subclass.

⚠️ Nits (code unchanged): `INHERITENCE` typo, stray `;` after `return self.__energy_type`, `wheeler_type=2` for a car, `new_car` is reused as a variable name (now a `Porsche`, earlier a `Car`).


```python
# INHERITENCE
# Just like Java, Python inheritence lets the child class use the parent class's methods and variables. 
# Only syntax changes here as well
# Note: We need to have getters first if we want to have setters in our class. Order Matters

class Vehicle:
    def __init__(self, name, wheeler_type, energy_type, horse_power):
        self.__name = name
        self.__wheeler_type = wheeler_type
        self.__energy_type = energy_type
        self.__horse_power = horse_power

    def revv(self):
        print(f"{self.name} says Vroom, Vroom")

    @property
    def name(self):
        return self.__name
    @name.setter
    def name(self, name):
        self.__name = name

    @property
    def wheeler_type(self):
        return self.__wheeler_type
    @wheeler_type.setter
    def wheeler_type(self, wheels):
        self.__wheeler_type = wheels
    
    @property
    def energy_type(self):
        return self.__energy_type;
    @energy_type.setter
    def energy_type(self, energy):
        self.__energy_type = energy

    @property
    def horse_power(self):
        return self.__horse_power
    @horse_power.setter
    def horse_power(self, hp):
        self.__horse_power = hp


class Porsche(Vehicle):      # (Vehicle) instead of extends...
    # If we're gonna inherit a class then the child class should absolutely inherit the parent class properties before instantiating parent class object
    def __init__(self, name, wheeler_type, energy_type, horse_power, turbo = True):
        #similar to super() in Java
        super().__init__(name, wheeler_type, energy_type, horse_power)
        self.__turbo = turbo


new_car = Porsche("Porsche GT 911", 2, "Petrol", 1200, True)
new_car.revv()
```

### Polymorphism and method overriding

**Takeaway:** same method call, different behaviour depending on the object's class.

- Overriding: a child redefines a parent method; Python picks the child's version at runtime.
- `Dog` has no `__init__`, so `Dog()` runs `Animal`'s constructor. `Cat` and `Husky` override `__init__` without `super().__init__()`, so `Animal`'s never runs for them.
- `super().method()` inside an override would extend the parent's behaviour instead of replacing it.
- No method overloading in Python; use default parameter values.
- Only `Dog` is instantiated here; `Cat` and `Husky` are defined but unused.


```python
# Polymorphism and Method Overriding

# Sadly, we dont have method overloading in Python which is similar to Java. If we want to use it, we need to have default 
# Parameter values

class Animal:
    def __init__(self):
        print("Animal Constructor")
    def make_some_noise(self):
        print("Animal Noises")
class Dog(Animal):
   
    def make_some_noise(self):
        print("Dog Barks")
class Cat(Animal):
    def __init__(self):             # Overwrites Animal __init__() constructor
        print("Cat Constructor")
    def make_some_noise(self):
        print("Cat meows")
class Husky(Dog):
    def __init__(self):
        print("Husky's constructor")
    def make_some_noise(self):
        print("Husky irritates")
dog = Dog()
dog.make_some_noise()
```

### Dataclasses + lambda sort

**Takeaway:** `@dataclass` generates `__init__`/`__repr__`/`__eq__` from the annotations; `asdict` turns an instance into a dict; `sort(key=lambda ...)` orders by a tuple key.

- Closest Java match is a `record`, but a dataclass is **mutable** unless `@dataclass(frozen=True)`.
- Annotations are hints only: `animal_weight: float` still accepts the int `200`.
- `sort` is **stable** and in place. The 2nd sort decides the order; the 1st (by `cost`) only breaks ties between equal kind and weight, so here it changes nothing.
- Key `(animal_kind, -animal_weight)` = kind ASC, weight DESC. Negating only works for numbers.
- Output `['Tiger', 'Lion', 'Elephant']`: Carnivores first (heavier Tiger before Lion), then Herbivore.
- Lambda 2: `max(zoo_dict, key=lambda x: x["cost"])` returns the whole dict with the highest cost (Elephant).
- Lambda 3: `check = lambda x: x%2==0`; `check(5)` → `False`. Prefer `def` when you name it (PEP 8).

⚠️ `say_something` calls itself → `RecursionError` if ever called (never called here). It should `print` a string, not `self.say_something()`.


```python
#Dataclasses -> Similar to records in Java and asdict 
#Lambdas

from dataclasses import dataclass, asdict

@dataclass
class Zoo:
    animal_name: str
    animal_kind: str
    animal_weight: float
    cost: float
    size_range: str
    def say_something(self):
        print(self.say_something())
z1 = Zoo("Lion", "Carnivore", 200, 80000, 'L')
z2 = Zoo("Tiger", 'Carnivore', 250, 79999, 'L')
z3 = Zoo('Elephant', 'Herbivore', 800.5, 100_000.80, 'XXXL')

zoo_dict = [asdict(z) for z in [z1,z2,z3]]

#--------------+-----------------+---------------+--------------#
#Lambda 1
zoo_dict.sort(key= lambda o : (o["cost"])) # Sort by cost ASC
zoo_dict.sort(key=lambda x : (x["animal_kind"], -x["animal_weight"])) # sort by animal_kind ASC, animal_weight DSC
print([n["animal_name"] for n in zoo_dict])

#Lambda 2
print(max(zoo_dict,key=lambda x : x["cost"]))  # Using a lamdba in an Aggregate function

 #Lambda 3
check = lambda x: x%2==0   
print(check(5))
```

### Quick review: mutability + lambda

- **Lambda 3:** `check = lambda x: x%2==0` is an anonymous one-expression function; `check(5)` → `False`. Prefer `def` when you assign it to a name (PEP 8).
- **Mutable:** list, dict, set. **Immutable:** tuple, str, int, float, frozenset. Only hashable (immutable) objects can be dict keys / set members.
- **List vs generator:** list stores everything (fast re-use, more memory); generator is lazy and single-use.
- **Interview traps from this notebook:** shallow vs deep copy, `==` vs `is`, `*args` tuple vs `**kwargs` dict, generator exhaustion, name mangling is not security.


### map, filter, reduce

**Takeaway:** `map(f, it)` transforms, `filter(f, it)` keeps matches, `functools.reduce(f, it)` folds to one value. `map`/`filter` return lazy iterators; wrap in `list()`.

- Plan: build a list of random ints, then apply the three.
- `reduce` lives in `functools` (not a builtin in Python 3).
- A comprehension or `sum()`/`max()` is often clearer than `reduce`.

⚠️ Bug: `random.randint(50)` raises `TypeError` (needs `a, b`), and an int is not iterable anyway. Use `[random.randint(1, 50) for _ in range(10)]`.


```python
# map, filter, reduce and zip
import random
from functools import reduce
lambda_list = [random.randint(10,50) for i in range(10)]

#map
mapped_list = list(map(lambda x : (x*2), lambda_list))

#filter
filtered_list = list(filter(lambda x : (x%5==0), lambda_list))
print(f"filtered list : {filtered_list}")

#zip
zipped_list = list(zip(lambda_list, mapped_list))
print(f"lambda list : {lambda_list}\nmapped list : {mapped_list}\nzipped list : {zipped_list}")

#reduce
reduced_value = reduce(lambda a,b : (a+b), lambda_list)
print(f"reduced value : {reduced_value}")
```
