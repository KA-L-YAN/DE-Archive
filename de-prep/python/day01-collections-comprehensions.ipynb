### Palindrome / anagram helpers
> **Known issues (code left unchanged):**
> - `check_palindrom(str1, str2)` checks whether `str1` is the reverse of `str2`. A real palindrome check takes **one** string and compares it with its own reverse. It is also never called.

> - This cell uses `Counter`, imported in the word-frequency cell above, so run that cell first.

```python
def check_palindrom(str1, str2):
    return (str1.lower() == str2[::-1].lower())

def check_anagrams(str1, str2):
    return Counter(str1.lower()) == Counter(str2.lower())

print(check_anagrams('mama', 'papa'))
print(Counter('mama'))
```

### `==` vs `is`
- `==` compares **values** (it calls `__eq__`).
- `is` compares **identity**: do both names point to the very same object?

Below, `a` and `b` hold equal values but are two separate lists, so `a == b` is `True` and `a is b` is `False`. After `a = b`, both names point to the same object, so `a is b` becomes `True`.

```python
a = [1,2,3,4]
b = [1,2,3,4]
print(a==b)
print(a is b)

a = b
print(a is b)
```

### Generators
A generator produces values **lazily**, one at a time, instead of storing them all in memory like a list. That makes it memory-efficient for large datasets.

**Interview answer:** A generator produces values lazily, one at a time, and is useful for memory-efficient processing of large datasets.

**Precisely:** a function that contains `yield` is a *generator function*. Calling it returns a *generator iterator*. `yield` hands back a value and pauses the function until the next value is requested.

**Watch out:** the `for` loop in the cell below calls `get_numbers()` again, which creates a **new** generator, so it starts from 1 again instead of continuing after the two `next()` calls. A generator is single-use: once exhausted, it yields nothing more.

```python
def get_numbers():
    yield 1
    yield 2
    yield 3
    yield 4
numbers  = get_numbers()
print(next(numbers))
print(next(numbers))

for i in get_numbers():
    print(i)
```

### Generator expressions
The official term (PEP 289) is **generator expression**, not "generator comprehension". It looks like a list comprehension but uses `()`, and it produces values lazily.

- Printing it only shows the generator object reference, not the values.
- Consume it with `next()`, a `for` loop, `list()`, `sum()`, etc. Note that `next` is a built-in **function**, not a keyword (the text printed in the cell below says "keyword").
- Here the `for` loop reuses the **same** generator, so it continues after the two `next()` calls and prints `5 7 9`.

```python
even_numbers = [i for i in range(0,11) if i%2==0]
print(f"Normal list using comprehension : {even_numbers}")

odd_numbers = (i for i in range(0,11) if i%2!=0)
print(f"Generator : {odd_numbers}\n"+
      "Yes this wont print anything other than the object reference... you have to use next keyword to print")

print(next(odd_numbers))
print(next(odd_numbers))
for i in odd_numbers:
    print(i, end = ' ')
```

### `==` calls `__eq__`
`a == b` calls `type(a).__eq__(a, b)`.
- If it returns `NotImplemented`, Python tries the reflected `type(b).__eq__(b, a)`.
- Exception: if `type(b)` is a **subclass** of `type(a)` and overrides `__eq__`, the right operand's method is tried **first**.
- If both return `NotImplemented`, `==` falls back to identity (`a is b`); it does not raise.
- Calling `__eq__` directly skips this fallback: `(5).__eq__(5.0)` gives `NotImplemented`, yet `5 == 5.0` is `True`.

```python
a = 5
b = 5
print(a.__eq__(b))
```

### Name mangling (the double underscore `__`)
**Purpose:** avoid accidental name clashes / overriding in **subclasses**. It is *not* a security feature.

Inside class `Car`, `self.__make` is rewritten at compile time to `self._Car__make`. The data is still reachable as `new_car._Car__make`, so privacy still runs on trust. Mangling applies only to names that start with two underscores and do **not** end with two (so `__init__` is untouched).

The read-only `@property` getters (`make`, `model`, ...) expose the mangled attributes safely; `on_road_price` also has a `@on_road_price.setter` meant to reject non-positive prices.

> **Known issues (code left unchanged):**
> - The setter line `print("Zero Value Error") if price <= 0 else self.__on_road_price = price` is a **SyntaxError**: an assignment is a statement and cannot be part of a conditional expression, so this cell will not run until it is rewritten as a normal `if / else` block. A setter should also `raise ValueError(...)` rather than `print`.
> - `get_details()` returns a **set** (comma-separated f-strings in `{ }` with no `key: value` pairs), so the output order is arbitrary. If a dict was intended, it needs pairs like `"Make": self.__make`.

```python
class Car:
    def __init__(self, make, model, manufactured_year, on_road_price):
        self.__make = make
        self.__model = model
        self.__manufactured_year = manufactured_year
        self.__on_road_price = on_road_price

    @property
    def make(self):
        return self.__make
    @property
    def model(self):
        return self.__model
    @property
    def manufactured_year(self):
        return self.__manufactured_year
    @property
    def on_road_price(self):
        return self.__on_road_price
    
    @on_road_price.setter
    def on_road_price(self, price):
        print("Zero Value Error") if price <= 0 else self.__on_road_price = price

    def get_details(self):
        return {f"Make : {self.__make}",
                f"Model : {self.__model}",
                f"Year : {self.__manufactured_year}",
                f"Price : {self.__on_road_price}"}

new_car = Car('Tesla', 'roadster', '2027', 200_000.00)
print(new_car.get_details())
```
