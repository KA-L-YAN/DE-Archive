```python
from collections import Counter
import string

string_list = "Hey there !!! Pavan is calling another Pavan and this another pavan is looking at another pavan"

def word_ferquency(strr):
    strr2 = strr.lower().translate(str.maketrans("","",string.punctuation))
    return Counter(strr2.split()).most_common(3)
word_ferquency(string_list)
```

```python
def check_palindrom(str1, str2):
    return (str1.lower() == str2[::-1].lower())

def check_anagrams(str1, str2):
    return Counter(str1.lower()) == Counter(str2.lower())

print(check_anagrams('mama', 'papa'))
print(Counter('mama'))
```

```python
# ==  compares the values
# is compares whether the 2 variables are pointing to a same object or not..

a = [1,2,3,4]
b = [1,2,3,4] # both are same in terms of values
print(a==b)
print(a is b) # but they dont point to the same object

a = b
print(a is b) # now they point to the same object
```

```python
# *args vs **kwargs
# *args - takes all the positional arguments as a tuple
# **kwargs - takes all the extra keyword arguments into a dictionary

def printer(*args, **kwargs):
    print(f"Type of *args : {type(args).__name__}\nvalues: {args}")
    print(f"Type of **kwargs : {type(kwargs).__name__}\nvalues: {kwargs}")

printer(1,2,3,3,4,'a','b','c', 2.14, name = "pavan", age = 59)
```

```python
# Generator -  Used to produce elements/values one at a time instead of storing all at once in the memory(like normal lists/collections)

# Interview answer: 
# A generator produces values lazily, one at a time, and is useful for memory-efficient processing of large datasets.

#Remember: A generator is the mechanism for producing values one at a time;
#  yield is the keyword that lets a function produce those values and pause.

def get_numbers():
    yield 1
    yield 2
    yield 3
    yield 4
numbers  = get_numbers()
print(next(numbers))
print(next(numbers))
# or

for i in get_numbers():
    print(i)
```

```python
# Generator comprehension

even_numbers = [i for i in range(0,11) if i%2==0]
print(f"Normal list using comprehension : {even_numbers}")

odd_numbers = (i for i in range(0,11) if i%2!=0)
print(f"Generator : {odd_numbers}\n"+
      "Yes this wont print anything other than the object reference... you have to use next keyword to print")

print(next(odd_numbers))
print(next(odd_numbers))
#or
for i in odd_numbers:
    print(i, end = ' ')
```
