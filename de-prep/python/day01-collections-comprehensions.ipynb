### Palindrome / anagram

- **Anagram:** same characters, same counts → `Counter(a) == Counter(b)`.
- ⚠️ `check_palindrom` compares `str1` with reverse of `str2`; a real palindrome check uses **one** string.
- ⚠️ `check_anagrams` counts spaces/punctuation (`'dormitory'` vs `'dirty room'` → `False`).
- Needs `Counter` from the word-frequency cell.


```python
def check_palindrom(str1, str2):
    return (str1.lower() == str2[::-1].lower())

def check_anagrams(str1, str2):
    return Counter(str1.lower()) == Counter(str2.lower())

print(check_anagrams('mama', 'papa'))
print(Counter('mama'))
```

### `==` vs `is`

- `==` → same **value** (calls `__eq__`)
- `is` → same **object** (identity)

Two equal lists: `==` True, `is` False. After `a = b`: `is` True.


```python
a = [1,2,3,4]
b = [1,2,3,4]
print(a==b)
print(a is b)

a = b
print(a is b)
```

### Generators

**Interview answer:** a generator produces values lazily, one at a time, so it is memory-efficient for large data.

- Function with `yield` = generator function; calling it returns a generator iterator.
- `yield` returns a value and pauses until the next one is asked for.
- ⚠️ Single-use. The `for` loop below calls `get_numbers()` again → a **new** generator, so it restarts at 1.


```python
def get_numbers():
    yield 1
    yield 2
    yield 3
    yield 4
numbers  = get_numbers()
print(next(numbers))
print(next(numbers))

for i in get_numbers():
    print(i)
```

### Generator expression

- `( ... )` instead of `[ ... ]`; lazy, one value at a time. Official term: _generator expression_ (not "generator comprehension").
- Printing shows only the object; consume with `next()`, `for`, `list()`, `sum()`.
- `next` is a built-in **function**, not a keyword.
- Here the `for` reuses the **same** generator, so it continues: `5 7 9`.


```python
even_numbers = [i for i in range(0,11) if i%2==0]
print(f"Normal list using comprehension : {even_numbers}")

odd_numbers = (i for i in range(0,11) if i%2!=0)
print(f"Generator : {odd_numbers}\n"+
      "Yes this wont print anything other than the object reference... you have to use next keyword to print")

print(next(odd_numbers))
print(next(odd_numbers))
for i in odd_numbers:
    print(i, end = ' ')
```

### `==` calls `__eq__`

`a == b` → `type(a).__eq__(a, b)`

- `NotImplemented` → try reflected `type(b).__eq__(b, a)`

- Both `NotImplemented` → falls back to identity, no error
- Direct call skips this: `(5).__eq__(5.0)` is `NotImplemented`, but `5 == 5.0` is `True`

### Name mangling (`__name`)

**Purpose:** avoid name clashes in **subclasses**, not security.

- `self.__make` in `Car` becomes `_Car__make` (compile time); still reachable as `obj._Car__make`.
- Only names starting with `__` and not ending with `__`.
- `@property` = read-only getter; `@x.setter` adds validation.

⚠️ Setter line is a **SyntaxError** (assignment inside a conditional expression). Use a normal `if/else` and `raise ValueError`.
⚠️ `get_details()` returns a **set** (no `key: value` pairs), so order is random.


```python
class Car:
    def __init__(self, make, model, manufactured_year, on_road_price):
        self.__make = make
        self.__model = model
        self.__manufactured_year = manufactured_year
        self.__on_road_price = on_road_price

    @property
    def make(self):
        return self.__make
    @property
    def model(self):
        return self.__model
    @property
    def manufactured_year(self):
        return self.__manufactured_year
    @property
    def on_road_price(self):
        return self.__on_road_price
    
    @on_road_price.setter
    def on_road_price(self, price):
        print("Zero Value Error") if price <= 0 else self.__on_road_price = price

    def get_details(self):
        return {f"Make : {self.__make}",
                f"Model : {self.__model}",
                f"Year : {self.__manufactured_year}",
                f"Price : {self.__on_road_price}"}

new_car = Car('Tesla', 'roadster', '2027', 200_000.00)
print(new_car.get_details())
```
